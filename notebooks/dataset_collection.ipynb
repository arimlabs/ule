```python
!pip install requests

import requests
from bs4 import BeautifulSoup

urls = [
    "http://web.archive.org/web/20250912200821/https://life.pravda.com.ua/columns/tri-skladniki-zmin-u-shkilniy-osviti-shcho-y-navishcho-zminyuye-mon-309656/",
    "http://web.archive.org/web/20260123203018/https://life.pravda.com.ua/health/chi-mozhna-pidlitkam-piti-kavu-i-yak-vona-na-nih-vplivaye-312797/",
    "http://web.archive.org/web/20260122062054/https://epravda.com.ua/biznes/yak-ukrajinskiy-biznes-spravlyayetsya-z-vidklyuchennyami-svitla-generatori-vitrati-problemi-816920/",
    "http://web.archive.org/web/20260119210248/https://epravda.com.ua/biznes/rinok-praci-ukrajini-2026-zarplati-popit-na-fahivciv-i-novi-vikliki-816608/",
    "http://web.archive.org/web/20260114115926/https://life.pravda.com.ua/health/2026/01/13/250928/"
]

texts = []
for url in urls:
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'html.parser')
    article = soup.find('div', class_='post_article_text')
    if article:
        texts.append(article.get_text(separator=' ', strip=True))

from datasets import Dataset
from pathlib import Path
import os
from google.colab import userdata
from huggingface_hub import login
login(token=userdata.get('HF_TOKEN'))
ds = Dataset.from_dict({"text": texts})
ds.push_to_hub("a-l-o/shortnews", private=True)
```
